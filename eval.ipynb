```python
import sys
sys.path.append('/home/dongjiayi/workbench/denoise/scConCRL')

import scanpy as sc
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "1"

from omegaconf import OmegaConf

from scConCRL.utils import *
from scConCRL.model import *
from scConCRL.data_processing import *
from scConCRL.eval import *
```

```python
opt_file = '/home/dongjiayi/workbench/denoise/scConCRL/config/demo_real_data_denoise.yaml'
opt = OmegaConf.load(opt_file)

opt.data.data_name = 'hPancreas'
```

```python
scdata = SCData(opt=opt.data)
scdata.load_data()
scdata.split_data(split_mode='no_test')
scdata.get_dataloader()
```

```python
input_dim = scdata.adata.X.shape[1]
```

```python
scdata.adata
```

```python
input_dim = scdata.adata.X.shape[1]
cov_dim = scdata.cov.shape[1]
output_path = os.path.join(opt.output_path, opt.data.data_name)
```

```python
mdl = model(
    input_dim = input_dim,
    domain_dim = cov_dim,
    hparams = opt.model,
)
mdl.load(output_path)  
```

```python
data = mdl.evaluate(
    dataloader=scdata.test_dataloader,
    data=scdata.adata,
)
```

```python
import matplotlib.pyplot as plt
color_palette =  ['#E24A33', '#F39B7F','#17BECF', '#348ABD', 
                '#988ED5', '#777777', '#FBC15E', 
                '#8EBA42', '#FFA5FF', '#8491B4']
plt.rcParams['text.color'] = 'black'
sc.set_figure_params(dpi=300, facecolor='white')
plt.style.use('ggplot')
plt.rcParams['figure.figsize'] = [8, 8]
```

```python
data.X = data.obsm['X_denoised']
sc.tl.pca(data, svd_solver="arpack")
sc.pp.neighbors(data, use_rep='X')
sc.tl.umap(data)

if opt.data.data_name == "hPBMC":
    color_list = ['cellstate','Method']
elif opt.data.data_name  == 'hHeart':
    color_list = ['cellstate','donor']
else: 
    color_list = ['cellstate','tech']

sc.pl.umap(data,
    color=color_list,
    palette = color_palette[:len(set(data.obs['cellstate']))],
    wspace=0.4,
    save=f'scConCRL_{opt.data.data_name}.pdf',
    legend_fontsize=16,
    size=40,
    frameon=True)
```
